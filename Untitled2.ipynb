```python
import pandas as pd

design_df = pd.read_csv("prompt_design_set.csv")
test_df = pd.read_csv("test_set.csv")
partial = pd.read_csv("test_set_all_conditions_results (2).csv")

TEXT_COL = "Consumer complaint narrative"
CATEGORIES = sorted(design_df["clean_category"].unique().tolist())

print("Categories:", CATEGORIES)
print("Test set:", test_df.shape, "| Already completed:", len(partial))
```

```python
!pip install openai -q

from openai import OpenAI
API_KEY = "paste your key here"

client = OpenAI(base_url="https://api.groq.com/openai/v1", api_key=API_KEY)
MODEL_NAME = "openai/gpt-oss-120b"
print("Client ready.")
```

```python
import time

def classify_zero_shot(complaint_text):
    prompt = (
        f"Classify this customer complaint into exactly one of these categories: {CATEGORIES}.\n\n"
        f'Complaint: "{complaint_text}"\n\n'
        f"Reply with ONLY the category name, nothing else."
    )
    start = time.time()
    response = client.chat.completions.create(
        model=MODEL_NAME, max_tokens=200, reasoning_effort="low",
        messages=[{"role": "user", "content": prompt}]
    )
    elapsed = time.time() - start
    return response.choices[0].message.content.strip(), elapsed, response.usage.total_tokens
```

```python
FEW_SHOT_EXAMPLES = design_df.groupby("clean_category").first().reset_index()

def build_few_shot_block():
    lines = []
    for _, row in FEW_SHOT_EXAMPLES.iterrows():
        lines.append(f'Complaint: "{row[TEXT_COL]}"\nCategory: {row["clean_category"]}\n')
    return "\n".join(lines)

FEW_SHOT_BLOCK = build_few_shot_block()

def classify_few_shot(complaint_text):
    prompt = (
        f"Here are some examples of customer complaints and their correct category:\n\n"
        f"{FEW_SHOT_BLOCK}\n"
        f"Now classify this new complaint into exactly one of these categories: {CATEGORIES}.\n\n"
        f'Complaint: "{complaint_text}"\n\n'
        f"Reply with ONLY the category name, nothing else."
    )
    start = time.time()
    response = client.chat.completions.create(
        model=MODEL_NAME, max_tokens=200, reasoning_effort="low",
        messages=[{"role": "user", "content": prompt}]
    )
    elapsed = time.time() - start
    return response.choices[0].message.content.strip(), elapsed, response.usage.total_tokens
```

```python
def classify_cot(complaint_text):
    prompt = (
        f"Classify this customer complaint into exactly one of these categories: {CATEGORIES}.\n\n"
        f'Complaint: "{complaint_text}"\n\n'
        f"First, think step by step about which category fits best and briefly explain your reasoning in 1-2 sentences. "
        f"Then, on a new final line, write EXACTLY: 'Final Category: <category name>'."
    )
    start = time.time()
    response = client.chat.completions.create(
        model=MODEL_NAME, max_tokens=300, reasoning_effort="low",
        messages=[{"role": "user", "content": prompt}]
    )
    elapsed = time.time() - start
    full_text = response.choices[0].message.content.strip()
    tokens_used = response.usage.total_tokens

    prediction = None
    for line in full_text.splitlines():
        if line.strip().lower().startswith("final category:"):
            prediction = line.split(":", 1)[1].strip()
    if prediction is None:
        prediction = full_text.strip().splitlines()[-1]

    return prediction, full_text, elapsed, tokens_used
```

```python
from collections import Counter

def classify_self_consistency(complaint_text, n_samples=5):
    predictions, explanations = [], []
    total_time, total_tokens = 0, 0
    for _ in range(n_samples):
        pred, expl, t, tok = classify_cot(complaint_text)
        predictions.append(pred)
        explanations.append(expl)
        total_time += t
        total_tokens += tok
    vote_counts = Counter(predictions)
    return vote_counts.most_common(1)[0][0], explanations, total_time, total_tokens, vote_counts
```

```python
def run_all_conditions(complaint_text, true_label, sc_samples=5):
    row_result = {"complaint": complaint_text, "true_label": true_label}

    pred, t, tok = classify_zero_shot(complaint_text)
    row_result.update({"zero_shot_pred": pred, "zero_shot_time": t, "zero_shot_tokens": tok})

    pred, t, tok = classify_few_shot(complaint_text)
    row_result.update({"few_shot_pred": pred, "few_shot_time": t, "few_shot_tokens": tok})

    pred, expl, t, tok = classify_cot(complaint_text)
    row_result.update({"cot_pred": pred, "cot_explanation": expl, "cot_time": t, "cot_tokens": tok})

    pred, expls, t, tok, votes = classify_self_consistency(complaint_text, n_samples=sc_samples)
    row_result.update({"sc_pred": pred, "sc_explanations": " ||| ".join(expls), "sc_time": t, "sc_tokens": tok, "sc_votes": dict(votes)})

    return row_result
```

```python
import pandas as pd
import os

# Check if 'partial' is defined, if not, attempt to load it
if 'partial' not in globals():
    partial_file = "test_set_all_conditions_results.csv"
    if os.path.exists(partial_file):
        try:
            partial = pd.read_csv(partial_file)
        except Exception as e:
            print(f"Warning: Could not load '{partial_file}'. Initializing partial as empty DataFrame. Error: {e}")
            partial = pd.DataFrame()
    else:
        print(f"Warning: '{partial_file}' not found. Initializing partial as empty DataFrame.")
        partial = pd.DataFrame()

print("Partial file has:", len(partial), "rows")
```

```python
completed_complaints = set(partial["complaint"])
remaining_df = test_df[~test_df[TEXT_COL].isin(completed_complaints)].reset_index(drop=True)
print(f"Remaining to process: {len(remaining_df)}")

test_results = partial.to_dict("records")

for i, row in remaining_df.iterrows():
    print(f"[{i+1}/{len(remaining_df)}] True: {row['clean_category']}")
    try:
        result = run_all_conditions(row[TEXT_COL], row["clean_category"])
        test_results.append(result)
    except Exception as e:
        print(f"  ERROR: {e}")
        break

    if (i + 1) % 20 == 0:
        pd.DataFrame(test_results).to_csv("test_set_all_conditions_results_PARTIAL.csv", index=False)
        print("  --- progress saved ---")

pd.DataFrame(test_results).to_csv("test_set_all_conditions_results.csv", index=False)
print(f"\nTotal so far: {len(test_results)} / {len(test_df)}")
```

```python
import pandas as pd

results = pd.read_csv("test_set_all_conditions_results.csv")
print("Total rows:", len(results))

def normalize(s):
    return str(s).strip().lower()

results["true_norm"] = results["true_label"].apply(normalize)

accuracy_summary = {}
for condition in ["zero_shot", "few_shot", "cot", "sc"]:
    pred_col = f"{condition}_pred"
    results[f"{condition}_correct"] = results[pred_col].apply(normalize) == results["true_norm"]
    acc = results[f"{condition}_correct"].mean()
    accuracy_summary[condition] = acc
    print(f"{condition:12s} accuracy: {acc:.1%}")
```

```python
from sklearn.metrics import classification_report

for condition in ["zero_shot", "few_shot", "cot", "sc"]:
    print(f"\n{'='*20} {condition.upper()} {'='*20}")
    print(classification_report(
        results["true_label"],
        results[f"{condition}_pred"],
        zero_division=0
    ))
```

```python
cost_time_summary = pd.DataFrame({
    "Condition": ["Zero-shot", "Few-shot", "CoT", "Self-consistency"],
    "Accuracy": [accuracy_summary["zero_shot"], accuracy_summary["few_shot"], accuracy_summary["cot"], accuracy_summary["sc"]],
    "Avg Tokens": [
        results["zero_shot_tokens"].mean(),
        results["few_shot_tokens"].mean(),
        results["cot_tokens"].mean(),
        results["sc_tokens"].mean(),
    ],
    "Avg Time (s)": [
        results["zero_shot_time"].mean(),
        results["few_shot_time"].mean(),
        results["cot_time"].mean(),
        results["sc_time"].mean(),
    ],
})
cost_time_summary["Accuracy"] = (cost_time_summary["Accuracy"] * 100).round(1)
cost_time_summary["Avg Tokens"] = cost_time_summary["Avg Tokens"].round(0)
cost_time_summary["Avg Time (s)"] = cost_time_summary["Avg Time (s)"].round(2)

print(cost_time_summary.to_string(index=False))
cost_time_summary.to_csv("summary_table.csv", index=False)
```
